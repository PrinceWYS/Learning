```python
import torch
import numpy as np
import sys
sys.path.append('..')
import d2dl
```

Load data

```python
batch_size = 256
train_iter, test_iter = d2dl.load_data_fashion_mnist(batch_size)
```

Define model parameters

```python
num_inputs, num_outputs, num_hiddens = 784, 10, 256

W1 = torch.tensor(np.random.normal(0, 0.01, (num_inputs, num_hiddens)), dtype=torch.float32)
b1 = torch.zeros(num_hiddens, dtype=torch.float32)
W2 = torch.tensor(np.random.normal(0, 0.01, (num_hiddens, num_outputs)), dtype=torch.float32)
b2 = torch.zeros(num_outputs, dtype=torch.float32)

params = [W1, b1, W2, b2]
for param in params:
    param.requires_grad_(True)
```

Define activation function

```python
def relu(x):
    return torch.max(input=x, other=torch.tensor(0.0))
```

Define model

```python
def net(X):
    H = relu(torch.matmul(X.view(-1, num_inputs), W1) + b1)
    output = torch.matmul(H, W2) + b2
    return output
```

Define loss function

```python
loss = torch.nn.CrossEntropyLoss()
```

Train

```python
num_epochs, lr = 5, 100.0

d2dl.train_ch3(net, train_iter, test_iter, loss, num_epochs, batch_size, params, lr)
```
